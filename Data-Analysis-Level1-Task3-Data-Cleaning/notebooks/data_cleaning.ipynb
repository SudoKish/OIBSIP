## 2. Missing Data Handling

Different strategies are used for missing values depending on the data type and business meaning of each column.

- **Item:** Missing item names will be filled with `"Unknown"` because the exact product cannot be reliably inferred from the available information.
- **Price Per Unit:** Missing numerical values will be replaced using the median. The median is less sensitive to extreme values than the mean and is appropriate for transactional price data.
- **Quantity:** Missing quantities will be replaced using the median because quantity is numerical and the median provides a robust estimate without being strongly affected by unusual transactions.
- **Total Spent:** Where `Price Per Unit` and `Quantity` are available, `Total Spent` will be calculated as `Price Per Unit × Quantity`. Any remaining missing values will be filled using the median.
- **Discount Applied:** Missing Boolean values will be filled using the mode because the column represents a categorical True/False indicator and the most frequent observed value provides a simple consistent replacement.

No rows will be deleted because the missing values can be handled without unnecessarily reducing the dataset size.

```python
df["Item"] = df["Item"].fillna("Unknown")

print("Missing Item values:", df["Item"].isnull().sum())
```

```python
price_median = df["Price Per Unit"].median()

df["Price Per Unit"] = df["Price Per Unit"].fillna(price_median)

print("Price Per Unit median used:", price_median)
print("Remaining missing values:", df["Price Per Unit"].isnull().sum())
```

```python
quantity_median = df["Quantity"].median()

df["Quantity"] = df["Quantity"].fillna(quantity_median)

print("Quantity median used:", quantity_median)
print("Remaining missing values:", df["Quantity"].isnull().sum())
```

```python
missing_total_before = df["Total Spent"].isnull().sum()

df["Total Spent"] = df["Price Per Unit"] * df["Quantity"]

print("Missing Total Spent before recalculation:", missing_total_before)
print("Missing Total Spent after recalculation:", df["Total Spent"].isnull().sum())
```

```python
discount_mode = df["Discount Applied"].mode()[0]

print("Discount Applied mode:", discount_mode)
```

```python
df["Discount Applied"] = df["Discount Applied"].fillna(discount_mode)

print("Remaining missing values:", df["Discount Applied"].isnull().sum())
```

```python
print("Missing values after treatment:")
print(df.isnull().sum())
```

## 3. Duplicate Removal

Duplicate records were checked using all columns in the dataset.

The initial inspection identified 0 duplicate rows. Therefore, no records were removed during this stage.

Keeping the original row count ensures that valid transactions are not unnecessarily discarded.

```python
# Check and remove duplicate rows

rows_before_duplicates = len(df)
duplicates_before = df.duplicated().sum()

df = df.drop_duplicates()

rows_after_duplicates = len(df)
duplicates_removed = rows_before_duplicates - rows_after_duplicates

print("Duplicate rows before cleaning:", duplicates_before)
print("Duplicate rows removed:", duplicates_removed)
print("Duplicate rows after cleaning:", df.duplicated().sum())
```

## 4. Data Standardisation

Categorical values were inspected for inconsistent capitalization, spelling, and formatting.

The dataset already uses consistent category labels such as `Cash`, `Credit Card`, `Digital Wallet`, `Online`, and `In-store`. Therefore, no category values require manual mapping.

As a preventive cleaning step, leading and trailing whitespace will be removed from text columns.

```python
# Standardize only actual text columns

text_columns = [
    "Transaction ID",
    "Customer ID",
    "Category",
    "Item",
    "Payment Method",
    "Location"
]

for column in text_columns:
    df[column] = df[column].str.strip()

print("Text standardisation completed successfully.")
```

## 5. Data Type Correction

Data types are corrected according to the meaning of each column.

- `Transaction ID` and `Customer ID` are kept as strings because they are identifiers.
- `Transaction Date` is converted from string to datetime.
- `Quantity` is converted from float to integer because transaction quantities are whole numbers.
- `Discount Applied` is converted from object to Boolean because it represents whether a discount was applied.
- `Price Per Unit` and `Total Spent` remain floating-point numerical values.

```python
# Correct data types

df["Quantity"] = df["Quantity"].astype("int64")

df["Transaction Date"] = pd.to_datetime(
    df["Transaction Date"],
    errors="coerce"
)

df["Discount Applied"] = df["Discount Applied"].astype("bool")

print("Data types corrected successfully.")
```

```python
# Verify corrected data types

print("Data types after correction:")
print(df.dtypes)
```

## 6. Outlier Detection and Handling

The Interquartile Range (IQR) method was used to identify potential outliers in the numerical columns.

The analysis found:
- `Price Per Unit`: 0 outliers
- `Quantity`: 0 outliers
- `Total Spent`: 60 outliers

All 60 `Total Spent` outliers had a value of 410. Although the IQR method classified these observations as statistical outliers, they represent valid transactions because the dataset contains a maximum price of 41 and a maximum quantity of 10, making 410 a valid possible transaction total.

Therefore, these observations were retained rather than removed or capped. This prevents the loss of legitimate high-value transactions.

```python
# IQR-based outlier detection

outlier_summary = []

for column in ["Price Per Unit", "Quantity", "Total Spent"]:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    outlier_summary.append({
        "Column": column,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers)
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary
```

```python
# Verify retained Total Spent outliers

total_spent_outliers = df[df["Total Spent"] > 403.5]

print("Retained Total Spent outliers:", len(total_spent_outliers))
print("Unique outlier values:", total_spent_outliers["Total Spent"].unique())
```
